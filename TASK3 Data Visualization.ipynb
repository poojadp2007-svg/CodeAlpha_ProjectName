```python
avg_salary = df.groupby("department")["salary"].mean()

plt.figure(figsize=(8,5))
plt.bar(avg_salary.index, avg_salary.values)

plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["salary"], bins=10)

plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Number of Employees")

plt.show()
```

```python
plt.figure(figsize=(9, 5))

plt.hist(
    df["age"],
    bins=10,
    color="skyblue",
    edgecolor="black",
    alpha=0.8
)

plt.title("Employee Age Distribution", fontsize=16, fontweight="bold")
plt.xlabel("Age (Years)", fontsize=12)
plt.ylabel("Number of Employees", fontsize=12)

plt.grid(axis="y", linestyle="--", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

plt.scatter(
    df["age"],
    df["salary"],
    alpha=0.6,
    color="purple",
    edgecolor="black"
)

plt.title("Age vs Salary", fontsize=16, fontweight="bold")
plt.xlabel("Age (Years)", fontsize=12)
plt.ylabel("Salary", fontsize=12)

plt.grid(True, linestyle="--", alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    df["age"],
    df["salary"],
    s=70,
    color="teal",
    alpha=0.65,
    edgecolors="white",
    linewidth=0.8
)

plt.title("Age vs Salary", fontsize=18, fontweight="bold", pad=15)
plt.xlabel("Age (Years)", fontsize=12)
plt.ylabel("Salary (₹)", fontsize=12)

plt.grid(
    True,
    linestyle="--",
    linewidth=0.6,
    alpha=0.25
)

plt.gca().spines["top"].set_visible(False)
plt.gca().spines["right"].set_visible(False)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.barplot(
    data=df,
    x="department",
    y="salary",
    errorbar=None
)

plt.title("Average Salary by Department", fontsize=16, fontweight="bold")
plt.xlabel("Department")
plt.ylabel("Average Salary (₹)")

plt.grid(axis="y", linestyle="--", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x="department"
)

plt.title("Number of Employees by Department", fontsize=16, fontweight="bold")
plt.xlabel("Department")
plt.ylabel("Number of Employees")

plt.grid(axis="y", linestyle="--", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
department_counts = df["department"].value_counts()

plt.figure(figsize=(8, 8))

plt.pie(
    department_counts,
    labels=department_counts.index,
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops={"width": 0.45, "edgecolor": "white"},
    textprops={"fontsize": 11}
)

plt.title(
    "Employee Distribution by Department",
    fontsize=17,
    fontweight="bold",
    pad=20
)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df,
    x="department",
    y="salary"
)

plt.title(
    "Salary Distribution by Department",
    fontsize=17,
    fontweight="bold",
    pad=15
)

plt.xlabel("Department", fontsize=12)
plt.ylabel("Salary (₹)", fontsize=12)

plt.grid(axis="y", linestyle="--", alpha=0.25)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))

corr = df[["employee_id", "salary", "age"]].corr()

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5,
    square=True
)

plt.title(
    "Correlation Heatmap",
    fontsize=17,
    fontweight="bold",
    pad=15
)

plt.tight_layout()
plt.show()
```

```python
joining_year = df["joining_date"].dt.year.value_counts().sort_index()

joining_year
```

```python
plt.figure(figsize=(10, 5))

sns.lineplot(
    x=joining_year.index,
    y=joining_year.values,
    marker="o",
    linewidth=2.5
)

plt.title(
    "Employee Joining Trend by Year",
    fontsize=17,
    fontweight="bold",
    pad=15
)

plt.xlabel("Joining Year", fontsize=12)
plt.ylabel("Number of Employees", fontsize=12)

plt.xticks(joining_year.index)
plt.grid(axis="y", linestyle="--", alpha=0.25)

plt.tight_layout()
plt.show()
```

```python
avg_salary = df.groupby("department")["salary"].mean()

print("Highest average salary:", avg_salary.idxmax(), "₹", round(avg_salary.max(), 2))
print("Lowest average salary:", avg_salary.idxmin(), "₹", round(avg_salary.min(), 2))
```
